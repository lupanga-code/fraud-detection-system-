# Load and Prepare Data

```python
import pandas as pd

# Load dataset
df = pd.read_csv("synthetic_dataset.csv")

# Convert timestamp to datetime
df['Timestamp'] = pd.to_datetime(df['Timestamp'])

# Extract time features
df['Hour'] = df['Timestamp'].dt.hour
df['Day'] = df['Timestamp'].dt.day
```

# Feature Engineering

```python
# Flag unusual hours (midnight–5am or late night >10pm)
df['UnusualHour'] = df['Hour'].apply(lambda x: 1 if x < 6 or x > 22 else 0)

# Transactions per minute per customer
df['TxnCountPerMinute'] = df.groupby(['CustomerID','Timestamp'])['TransactionID'].transform('count')

# Flag multiple accounts at same time
df['MultiAccountFlag'] = df['TxnCountPerMinute'].apply(lambda x: 1 if x > 1 else 0)

# Location mismatch (compare with previous transaction location)
df['PrevLocation'] = df.groupby('CustomerID')['Location'].shift()
df['LocationMismatch'] = (df['Location'] != df['PrevLocation']).astype(int)

# Channel switching (compare with previous channel)
df['PrevChannel'] = df.groupby('CustomerID')['Channel'].shift()
df['ChannelSwitch'] = (df['Channel'] != df['PrevChannel']).astype(int)
```

# Encode Categorical Features

```python
from sklearn.preprocessing import LabelEncoder

# Encode categorical columns
for col in ['Merchant','Location','Channel']:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
```

# Define Features and Labels

```python
features = ["Amount", "Merchant", "Location", "Channel", "Hour"]
X = df[features]
y = df["FraudLabel"]   # au jina sahihi la column yako
```

# Train-Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

# Visualize Class Distribution Before SMOTE

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Plot class distribution before SMOTE
sns.countplot(x=y_train, palette="coolwarm")
plt.title("Fraud vs Genuine Transactions (Before SMOTE)")
plt.xlabel("FraudLabel (0 = Genuine, 1 = Fraud)")
plt.ylabel("Number of Transactions")
plt.show()
```

# Apply SMOTE on Training Data

```python
# Apply SMOTE only on training set
smote = SMOTE(random_state=42)

X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

print("Before SMOTE:", y_train.value_counts())
print("After SMOTE:", y_train_resampled.value_counts())
```

# Visualize Class Distribution After SMOTE

```python
# Plot class distribution after SMOTE
sns.countplot(x=y_train_resampled, palette="coolwarm")
plt.title("Fraud vs Genuine Transactions (After SMOTE)")
plt.xlabel("FraudLabel (0 = Genuine, 1 = Fraud)")
plt.ylabel("Number of Transactions")
plt.show()
```

# Train Model

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=200, random_state=42)
model.fit(X_train_resampled, y_train_resampled)
```

# Evaluate Model

from sklearn.metrics import classification_report, confusion_matrix

y_pred = model.predict(X_test)

print("Classification Report:\n", classification_report(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))

# Save Model

```python
import joblib

joblib.dump(model, "fraud_detection_model_simple.pkl")
```
